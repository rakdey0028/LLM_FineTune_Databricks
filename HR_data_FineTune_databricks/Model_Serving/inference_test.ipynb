```python
# Test both inference methods with the same question
print("="*80)
print("TESTING BOTH INFERENCE METHODS")
print("="*80)

test_question = "how many days of casual leave I can take max in one month?"

try:
    inference = DatabricksInference("hr-policy-assistant_instruct")
    
    print("\n📌 Question:")
    print(f"   {test_question}")
    
    print("\n" + "="*80)
    print("Method 1: LangChain (inference_generate)")
    print("="*80)
    result1 = inference.inference_generate(test_question)
    print(f"Answer: {result1}")
    
    print("\n" + "="*80)
    print("Method 2: MLflow Direct API (inference_generate_direct_api)")
    print("="*80)
    result2 = inference.inference_generate_direct_api(test_question)
    print(f"Answer: {result2}")
    
    print("\n" + "="*80)
    print("✅ Both methods working!")
    print("="*80)
    
    # Check if answers mention the wrong criteria
    if "one year" in result1.lower() or "one year" in result2.lower():
        print("\n⚠️  WARNING: Answer mentions 'one year' which is incorrect!")
        print("   The model needs to be re-deployed with chat_template fix.")
        print("   Go to model_deploy_serving notebook, Cell 8 to re-deploy.")
    
except Exception as e:
    print(f"\n❌ Error: {e}")
    import traceback
    traceback.print_exc()
```
